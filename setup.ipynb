```python
# This set up assumes that there is only one year's use.
# It does not consider having multiple year's of data.
from datetime import date
import os
import json

base_path = '/content/drive/MyDrive/Lesson Plan Helper'
meta_path = '/content/drive/MyDrive/Lesson Plan Helper/.metadata'

def mount_drive():
    """ Used to mount Google Drive. Does nothing if it is already mounted."""
    """ It will ask for permission to access. """
    if not os.path.exists('/content/drive'):
        from google.colab import drive
        drive.mount('/content/drive')


def calendar_exists():
    return os.path.exists('base_path') and os.path.exists('meta_path')

def create_calendar():
    if not calendar_exists():
        # create necessary directories
        os.makedirs(base_path, exist_ok=True)
        os.makedirs(meta_path, exist_ok=True)

        # create files to store meta data in
        os.chdir(meta_path)
        with open('days_off.json', 'w') as file:
            pass
        with open('lessons_and_standards.json', 'w') as file:
            pass

def get_days_off():
    """ Returns an empty list if ../days_off.json does not exist."""
    data = []
    if os.path.exists(meta_path):
        json_path = meta_path + '/days_off.json'
        if os.path.getsize(json_path) != 0:
            with open(json_path, "r") as file:
                data = json.load(file)
    return data

def save_days_off(days_off_list):
    """ Saves an entire list of days off. Not for appending to."""
    with open(meta_path+'/days_off.json', "w") as file:
        json.dump(days_off_list, file, indent=4)



```
